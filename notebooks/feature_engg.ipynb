```python
import pandas as pd

# === Load cleaned data ===
df_with = pd.read_csv("cleaned_linkedin_posts.csv")

# === Parse date again (if needed) ===
df_with['created_date'] = pd.to_datetime(df_with['created_date'], errors='coerce')

# Create new features

# 1. Total engagement
df_with['total_engagement'] = df_with['likes'] + df_with['comments'] + df_with['reposts']

# 2. Day of week
df_with['day_of_week'] = df_with['created_date'].dt.day_name()

# 3. Hour of post (if timestamp includes time — optional)
df_with['hour'] = df_with['created_date'].dt.hour

# 4. Month
df_with['month'] = df_with['created_date'].dt.month

# 5. Caption length
df_with['caption_length_words'] = df_with['post_title'].astype(str).apply(lambda x: len(x.split()))
df_with['caption_length_chars'] = df_with['post_title'].astype(str).apply(len)


# CTR bucket (optional)
df_with['ctr_bucket'] = pd.qcut(df_with['click_through_rate_(ctr)'], q=4, labels=["Low", "Med-Low", "Med-High", "High"])

# 8. Post type cleanup (standardize categories)
df_with['post_type'] = df_with['post_type'].str.strip().str.lower()

# === Save enhanced data ===
df_with.to_csv("feature_engineered_posts.csv", index=False)
print(" Feature-engineered data saved to feature_engineered_posts.csv")
```
