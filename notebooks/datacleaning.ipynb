```python
import pandas as pd
import os

df=pd.read_excel("WiTH2024june-2025jul.xls")

df.head()
```

```python
df_with=pd.read_csv("2425With.csv")

df_with.head()
```

```python
df_with.info()
```

```python
# === Standardize column names ===
df_with.columns = df_with.columns.str.strip().str.lower().str.replace(" ", "_")

# === Drop irrelevant or empty columns ===
drop_cols = ['campaign_name', 'campaign_start_date', 'campaign_end_date',
             'offsite_views', 'follows', 'content_type']
df_with = df_with.drop(columns=drop_cols, errors='ignore')

# === Parse date ===
df_with['created_date'] = pd.to_datetime(df_with['created_date'], errors='coerce')

# === Drop rows with missing post content or date ===
df_with = df_with.dropna(subset=['post_title', 'created_date'])

# === Drop duplicates ===
df_with = df_with.drop_duplicates()

# === Preview clean shape ===
print(f"Cleaned dataset shape: {df_with.shape}")

# === Save cleaned data ===
df_with.to_csv("cleaned_linkedin_posts.csv", index=False)
print("Saved to cleaned_linkedin_posts.csv")
```
